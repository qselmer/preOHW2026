## 1. Conjunto de datos

Se utiliza el archivo **`datos/datos_ejemplo.csv`**, incluido en este repositorio. Contiene seis registros mensuales de temperatura superficial del mar (°C) y concentración de clorofila (mg/m³).

**Importante:** son datos de ejemplo para practicar lectura y exploración; no deben interpretarse como observaciones oceanográficas reales ni atribuirse a una zona geográfica específica.

Se utiliza el módulo `csv`, incluido en Python, para leer el archivo sin instalar paquetes adicionales. La ruta relativa admite ejecutar Jupyter desde la carpeta de la actividad o desde la raíz del repositorio.

```python
from pathlib import Path
import csv

archivo = Path('datos/datos_ejemplo.csv')
if not archivo.is_file():
    archivo = Path('01_actividad/datos/datos_ejemplo.csv')

with archivo.open('r', encoding='utf-8-sig', newline='') as f:
    lector = csv.DictReader(f)
    columnas = lector.fieldnames
    datos = list(lector)

print(f'Archivo cargado: {archivo}')
print(f'Dimensiones: {len(datos)} filas y {len(columnas)} columnas')
```

## 3. Primeras filas

Se muestran las primeras cinco observaciones del conjunto de datos.

```python
print(' | '.join(columnas))
print('-' * 45)
for fila in datos[:5]:
    print(' | '.join(fila[col] for col in columnas))
```

## 4. Descripción de las variables

La **unidad de observación** es un mes del conjunto de ejemplo.

| Variable | Descripción | Unidad |
|---|---|---|
| `mes` | Mes correspondiente a cada registro | Categórica |
| `temperatura_c` | Temperatura superficial del mar de ejemplo | °C |
| `clorofila_mg_m3` | Concentración de clorofila de ejemplo | mg/m³ |

El archivo incluye **6 observaciones** (enero a junio) y **3 variables**.

```python
from statistics import mean, stdev

print('Número de observaciones:', len(datos))
print('Variables:', ', '.join(columnas))
print('Valores faltantes por variable:')
for col in columnas:
    faltantes = sum(fila[col].strip() == '' for fila in datos)
    print(f'  {col}: {faltantes}')

print('\nEstadísticas descriptivas (valores de ejemplo):')
for col in ('temperatura_c', 'clorofila_mg_m3'):
    valores = [float(fila[col]) for fila in datos if fila[col].strip()]
    print(f'{col}: n={len(valores)}, media={mean(valores):.2f}, '
          f'mínimo={min(valores):.2f}, máximo={max(valores):.2f}, '
          f'desviación estándar={stdev(valores):.2f}' if len(valores) > 1
          else f'{col}: n={len(valores)}')
```
